```python
#buiuld a calculator
num1 = int(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /): ")
num2 = int(input("Enter second number: "))

if operator == "+":
    result = num1 + num2
    print("Result:",num1+num2)

elif operator == "-":
    result = num1 - num2
    print("Result:",num1-num2)

elif operator == "*":
    result = num1 * num2
    print("Result:",num1*num2)

elif operator == "/":
    # Basic check to prevent crash when dividing by zero
    if num2 == 0:
        print("Error! Cannot divide by zero.")
    else:
        result = num1 / num2
        print("Result:",num1/num2)

else:
    print("Invalid operator! Please use +, -, *, or /.")
```

```python
#to calculate the cube of given number
cube_list = []

print("Please enter 5 numbers:")

for i in range(5):
    num = int(input(f"Enter number {i + 1}: "))
    
    cube_value = num * num * num
    
    cube_list.append(cube_value)

print("\nThe cube values are:", cube_list)
```

```python
#to reverse the order of given variable

num = int(input("Enter a number to reverse: "))

original_num = num
reversed_num = 0

while num > 0:
    last_digit = num % 10               
    reversed_num = (reversed_num * 10) + last_digit  
    num = num // 10                      

# 3. Display the result
print("Original Number:",original_num)
print("Reversed Number:",reversed_num)
```

```python
#to determine wheather the given year is leap year or not
year = int(input("Enter a year: "))

if year % 400 == 0:
    print("this is a leap year")
elif year % 100 == 0:
    print("this is a leap year")
elif year % 4 == 0:
    print("this is a leap year")
else:
    print("this is not a leap year")
```

```python
#Q:university grades using float command
marks = float(input("Enter Marks obtained in Math: "))

if marks >= 90 and marks <= 100:
    print("grade=O")
elif marks >= 80 and marks <= 89:
    print("grade=A+")
elif marks >= 70 and marks <= 79:
    print("grade=A")
elif marks >= 60 and marks <= 69:
    print("grade=B")
elif marks >= 50 and marks <= 59:
    print("grade=C")
elif marks >= 40 and marks <= 49:
    print("grade=P")
   
else:
    print("grade=F")

```

```python
#Q:university grades 
marks = int(input("Enter Marks obtained in Math: "))

if marks >= 90 and marks <= 100:
    grade = "O"
elif marks >= 80 and marks <= 89:
    grade = "A+"
elif marks >= 70 and marks <= 79:
    grade = "A"
elif marks >= 60 and marks <= 69:
    grade = "B"
elif marks >= 50 and marks <= 59:
    grade = "C"
elif marks >= 40 and marks <= 49:
    grade = "P"
elif marks >= 0 and marks <= 39:
    grade = "F"
else:
    grade = "Invalid Marks! Please enter a value between 0 and 100."

print("Grade Assigned:", grade)
```
