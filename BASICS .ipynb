```python
#take 3 sub marks,print %age,maximum marks in subject

math=float(input("enter math marks : "))
phy=float(input("enter phy marks : "))
chem=float(input("enter chem marks: "))

print("maths marks is: ",math)
print("phy marks is: ",phy)
print("chem marks is : ",chem)

percentage=(math+phy+chem)/3

print("percentage is : ",percentage)

if math>=chem and math>=phy:
    print("math has highest marks")
elif phy>=math and phy>=chem:
    print("phy has highest marks")
else:
    print("chem has highest marks")
```
